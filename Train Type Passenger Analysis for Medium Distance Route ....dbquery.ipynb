```sql
SELECT
DISTINCT
 rtc.Train_Type,
  rtc.Source_Station,
  rtc.Destination_Station,
 rtc.Number_of_Stations,
 rtc.Travel_Distance,
  COUNT(Train_Number) AS trains_count,
  SUM(Number_of_Passengers) AS total_passengers
FROM railway_ticket_confirmation rtc
WHERE rtc.Travel_Distance BETWEEN 500 and 1000
GROUP BY rtc.Source_Station, rtc.Destination_Station, rtc.Train_Type,rtc.Number_of_Stations,rtc.Travel_Distance
ORDER BY total_passengers DESC;
```
