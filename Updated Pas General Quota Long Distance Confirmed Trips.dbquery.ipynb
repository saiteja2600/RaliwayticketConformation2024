```sql
SELECT
rfc.Train_Number,
    rfc.Date_of_Journey,
    rfc.Booking_Date,
    rfc.Source_Station,
    rfc.Destination_Station,
    rfc.Travel_Distance,
    rfc.Class_of_Travel,
    rfc.Age_of_Passengers AS age_category,
    SUM(rfc.Number_of_Passengers) AS total_passengers,
    rfc.Confirmation_Status
    
FROM railway_ticket_confirmation rfc
WHERE rfc.Quota = 'General'
  AND rfc.Travel_Distance > 1000
  and rfc.Date_of_Journey='2106-10-21'
  
GROUP BY
rfc.Train_Number,
    rfc.Date_of_Journey,
    rfc.Booking_Date,
    rfc.Source_Station,
    rfc.Destination_Station,
    rfc.Travel_Distance,
    rfc.Class_of_Travel,
    rfc.Age_of_Passengers,
    rfc.Confirmation_Status
ORDER BY
    rfc.Date_of_Journey,
    total_passengers DESC;
```
