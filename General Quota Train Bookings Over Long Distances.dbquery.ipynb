```sql
SELECT
   rfc.Train_Number,rfc.Class_of_Travel, rfc.Date_of_Journey,rfc.Number_of_Passengers,rfc.Source_Station,rfc.Destination_Station,rfc.Booking_Date
    
FROM railway_ticket_confirmation rfc
WHERE rfc.Quota = 'General'
  AND rfc.Travel_Distance > 1000 and rfc.Confirmation_Status = 'Confirmed'
GROUP BY rfc.Date_of_Journey, rfc.Train_Number, rfc.Number_of_Passengers,rfc.Source_Station,rfc.Destination_Station,rfc.Booking_Date,rfc.Class_of_Travel
ORDER BY rfc.Date_of_Journey;
```
