```sql
SELECT DISTINCT
    rfc.Train_Number,
    rfc.Date_of_Journey,
    rfc.Booking_Date,
    rfc.Source_Station,
    rfc.Destination_Station,
    rfc.Travel_Distance,
    rfc.Class_of_Travel,
    rfc.Age_of_Passengers,
    rfc.Booking_Channel
FROM railway_ticket_confirmation rfc
WHERE year(rfc.Date_of_Journey) BETWEEN 2024 AND year(current_date()) and month(rfc.Date_of_Journey) BETWEEN 1 AND month(current_date())
  AND rfc.Travel_Distance > 1000
ORDER BY rfc.Date_of_Journey DESC;
```
