```python
railway_df_rename = railway_df.withColumnRenamed("PNR Number","PNR_Number") \
.withColumnRenamed("Train Number","Train_Number") \
.withColumnRenamed("Date of Journey","Date_of_Journey") \
.withColumnRenamed("Class of Travel","Class_of_Travel") \
.withColumnRenamed("Quota","Quota") \
.withColumnRenamed("Source Station","Source_Station") \
.withColumnRenamed("Destination Station","Destination_Station") \
.withColumnRenamed("Booking Date","Booking_Date") \
.withColumnRenamed("Boarding Station","Boarding_Station") \
.withColumnRenamed("Route","Route") \
.withColumnRenamed("Current Status","Current_Status") \
.withColumnRenamed("Number of Passengers","Number_of_Passengers") \
.withColumnRenamed("Age of Passengers","Age_of_Passengers") \
.withColumnRenamed("Booking Channel","Booking_Channel") \
.withColumnRenamed("Travel Distance","Travel_Distance") \
.withColumnRenamed("Number of Stations","Number_of_Stations") \
.withColumnRenamed("Travel Time","Travel_Time") \
.withColumnRenamed("Train Type","Train_Type") \
.withColumnRenamed("Seat Availability","Seat_Availability") \
.withColumnRenamed("Special Considerations","Special_Considerations") \
.withColumnRenamed("Holiday or Peak Season","Holiday_or_Peak_Season") \
.withColumnRenamed("Waitlist Position","Waitlist_Position") \
.withColumnRenamed("Confirmation Status","Confirmation_Status")
display(railway_df_rename)
```

```python
railway_df=railway_df_rename.write.mode("overwrite").format("delta").option("overwriteSchema","true").saveAsTable("datastore2026.railwayticket2024.railway_ticket_confirmation")
```
